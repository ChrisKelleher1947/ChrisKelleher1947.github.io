```python
import librosa
import matplotlib.pyplot as plt

def plot_spectrum(file, title):
    audio, sr = librosa.load(file, sr=None)
    spectrum = abs(librosa.stft(audio))

    plt.figure()
    plt.title(title)
    plt.imshow(librosa.amplitude_to_db(spectrum), aspect='auto', origin='lower')
    plt.colorbar()
    plt.show()

plot_spectrum("tts_output.mp3", "Normal Audio")
plot_spectrum("voice-1775734317128.ogg", "WhatsApp Audio")
```
